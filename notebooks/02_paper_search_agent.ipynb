```python
import sys
from pathlib import Path

project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from langchain_ollama import ChatOllama

from src.paper_search_agent import PaperSearchAgent
from src.paper_tools import PaperTools

model = ChatOllama(model="llama3.2:latest", temperature=0, num_ctx=8192)
paper_tools = PaperTools()
paper_search_agent = PaperSearchAgent(model, paper_tools)
```

```python
print(model.invoke("Say hello to my students in one short sentence.").content)
```

```python
request = "Find two arXiv papers about large language models and show their links."
result = paper_search_agent.invoke(request)
```

```python
for message in result["messages"]:
    message.pretty_print()
```
