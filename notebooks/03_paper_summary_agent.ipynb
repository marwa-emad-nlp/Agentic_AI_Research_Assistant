```python
import sys
from pathlib import Path

project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from langchain_ollama import ChatOllama
from src.paper_summary_agent import PaperSummaryAgent

model = ChatOllama(model="llama3.2:latest", temperature=0, num_ctx=8192)
paper_summary_agent = PaperSummaryAgent(model)
```

```python
request = """Summarize this fictional paper in two short sentences.

Title: Query Reformulation for Academic Search
Abstract: We studied whether rewriting search queries improves academic
paper discovery. We compared a single keyword search with a workflow that
could reformulate the query once. On 20 test questions, the reformulation
workflow found at least one relevant paper in the top five results for
15 questions, compared with 12 for the single-search workflow.
"""

result = paper_summary_agent.invoke(request)
print(result["messages"][-1].content)
```
